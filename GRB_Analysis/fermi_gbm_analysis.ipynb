```python
from heapy.data.retrieve import gbmRetrieve
from heapy.geos.geometry import gbmGeometry
from heapy.pipe.event import gbmTTE
from heapy.util.time import fermi_utc_to_met
from heapy.util.tools import format_message
```

```python
utc = '2024-02-19T06:21:42'
fermi_met = fermi_utc_to_met(utc)
```

```python
gbm_rtv = gbmRetrieve.from_utc(utc=utc, t1=-500, t2=500)
gbm_rtv.rtv_res
```

```python
gbm_geo = gbmGeometry(file=gbm_rtv.rtv_res['poshist'])
```

```python
gbm_geo.get_saa_passage(met=[fermi_met - 500, fermi_met, fermi_met + 500])
```

```python
gbm_geo.get_location_visible(
    ra=80.016, dec=25.541, met=[fermi_met - 500, fermi_met, fermi_met + 500]
)
```

```python
dets = ['n0', 'n1', 'n2', 'n3', 'n4', 'n5', 'n6', 'n7', 'n8', 'n9', 'na', 'nb', 'b0', 'b1']

for det in dets:
    angle = gbm_geo.get_detector_angle(
        ra=80.016, dec=25.541, det=det, met=[fermi_met - 100, fermi_met, fermi_met + 100]
    )
    print(det, angle)
```

```python
gbm_geo.extract_skymap(ra=80.016, dec=25.541, met=fermi_met, savepath='./geometry')
```

```python
for det in ['n9', 'na', 'b1']:
    gbm_tte = gbmTTE(gbm_rtv.rtv_res['tte'][det])
    gbm_tte.filter_time([-200, 300], fermi_met)
    if det[0] == 'n':
        gbm_tte.filter_energy([8, 900])
    else:
        gbm_tte.filter_energy([300, 38000])

    print(format_message(f'extract light curve for {det}'))

    gbm_tte.lc_binsize = 2
    gbm_tte.bs_ignore = [[0, 100]]
    gbm_tte.extract_curve(savepath=f'./curve/{det}', show=True)
```

```python
for det in ['n9', 'na']:
    gbm_tte = gbmTTE(gbm_rtv.rtv_res['tte'][det])
    gbm_tte.filter_time([-200, 300], fermi_met)
    if det[0] == 'n':
        gbm_tte.filter_energy([8, 900])
    else:
        gbm_tte.filter_energy([300, 38000])

    gbm_tte.lc_binsize = 2
    gbm_tte.bs_ignore = [[0, 100]]
    gbm_tte.calculate_txx(xx=0.9, savepath=f'./curve/{det}/t90')
```

```python
for det in ['n9', 'na', 'b1']:
    gbm_tte = gbmTTE(gbm_rtv.rtv_res['tte'][det], gbm_rtv.rtv_res['poshist'])
    gbm_tte.filter_time([-400, 400], fermi_met)

    print(format_message(f'extract spectrum and response for {det}'))

    gbm_tte.spec_slices = [[0, 70]]
    gbm_tte.extract_spectrum(savepath=f'./spectrum/{det}', show=True)
    gbm_tte.extract_response(ra=80.016, dec=25.541, savepath=f'./spectrum/{det}')
```

```python
import numpy as np

from bayspec import BayesInfer, Data, DataUnit, Plot
from bayspec.model.local import *
```

```python
import sys
sys.path.insert(1,'/workspace/GRB_Analysis/quickstart')
savepath = '/workspace/GRB_Analysis/quickstart'
```

```python
nai = DataUnit(
    src='./ME/me.src',
    bkg='./ME/me.bkg',
    rsp='./ME/me.rsp',
    notc=[8, 900],
    stat='pgstat',
    rebn={'min_sigma': 2, 'max_bin': 10},
)

bgo = DataUnit(
    src='./HE/he.src',
    bkg='./HE/he.bkg',
    rsp='./HE/he.rsp',
    notc=[300, 38000],
    stat='pgstat',
    rebn={'min_sigma': 2, 'max_bin': 10},
)

data = Data([('nai', nai), ('bgo', bgo)])

data.save(savepath)
data
```

```python
model = cpl()
model.save(savepath)
model
```

```python
infer = BayesInfer([(data, model)])
infer.save(savepath)
infer
```

```python
post = infer.multinest(nlive=400, resume=True, verbose=False, savepath=savepath)
post.save(savepath)
post
```

```python
fig = Plot.infer(post, style='CE')
fig.save(f'{savepath}/ctsspec')
```

```python
fig = Plot.infer(post, style='NE')
fig.save(f'{savepath}/phtspec')
```

```python
fig = Plot.post_corner(post, ploter='plotly')
fig.save(f'{savepath}/corner')
```

```python
earr = np.logspace(1, 3, 100)

modelplot = Plot.model(ploter='plotly', style='vFv', post=True)
modelplot.add_model(model, E=earr)
fig = modelplot.get_fig()
fig.save(f'{savepath}/model')
```
