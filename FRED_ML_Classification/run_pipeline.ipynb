```python
import importlib
import fred_scorer
importlib.reload(fred_scorer)
from fred_scorer import score_all_bursts, plot_best_per_grb_by_tier, plot_all_detectors_by_grb

# Score all 53 GRBs
df = score_all_bursts('/workspace/data')

# Plot 1: Best detector per GRB, organized by tier
figs_tier, best_per_grb = plot_best_per_grb_by_tier(df, ncols=5)
for tier_name, fig in figs_tier:
    safe_name = tier_name.replace(' ', '_').lower()
    fig.savefig(f'/workspace/data/fred_tier_{safe_name}.png', dpi=120)
    print(f"Saved: fred_tier_{safe_name}.png")

# Plot 2: All detectors per GRB (shows variation across detectors)
fig_all = plot_all_detectors_by_grb(df, ncols=5)
fig_all.savefig('/workspace/data/fred_all_detectors_per_grb.png', dpi=120)
print(f"Saved: fred_all_detectors_per_grb.png")

# Summary stats
print(f"\n=== SCORING SUMMARY ===")
print(f"Total GRBs: {len(best_per_grb)}")
print(f"Excellent (>={best_per_grb['fred_score'].quantile(0.33):.2f}): {(best_per_grb['fred_score'] >= best_per_grb['fred_score'].quantile(0.33)).sum()}")
print(f"Moderate: {((best_per_grb['fred_score'] >= best_per_grb['fred_score'].quantile(0.67)) & (best_per_grb['fred_score'] < best_per_grb['fred_score'].quantile(0.33))).sum()}")
print(f"Poor: {(best_per_grb['fred_score'] < best_per_grb['fred_score'].quantile(0.67)).sum()}")
print(f"\nTop 10 FRED candidates:\n{best_per_grb.head(10)[['grb_name', 'det', 'fred_score']]}")
```
